```python
calls_per_agent = df_call['agent_id'].value_counts().reset_index()
calls_per_agent.columns = ['agent_id', 'count']

mean_calls = calls_per_agent['count'].mean().round(2)
calls_per_agent['mean'] = mean_calls

calls_per_agent['diff'] = calls_per_agent['count'] - mean_calls

calls_per_agent.sort_values('diff', ascending=False)
```
